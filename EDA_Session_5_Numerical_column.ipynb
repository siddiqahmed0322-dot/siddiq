**np.where**

**capping**

- we can fill the  outliers which is less than 25P with 25P value

- we can fill the  outliers which is less than 75P with 75P value

```python
wage_data=visa['prevailing_wage']
q1=np.percentile(wage_data,25)
q2=np.percentile(wage_data,50)
q3=np.percentile(wage_data,75)
IQR=q3-q1
lb=q1-1.5*IQR
ub=q3+1.5*IQR
visa['wage_data_cap']=wage_data.clip(lb,ub)
```

```python
wage_data=visa['prevailing_wage']
wage_data_q2=visa['wages_q2']
wage_data_cap=visa['wage_data_cap']
plt.subplot(1,3,1).boxplot(wage_data,vert=False)
plt.subplot(1,3,2).boxplot(wage_data_q2,vert=False)
plt.subplot(1,3,3).boxplot(wage_data_cap,vert=False)

plt.show()
```
