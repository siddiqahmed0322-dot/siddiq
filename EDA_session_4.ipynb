```python
keys=visa['continent'].value_counts().keys()
values=visa['continent'].value_counts().values
# plt.figure(figsize=())
plt.pie(values,
        explode=[0.1 for i in range(6)],
       labels=keys,
        radius=0.8,
       autopct='%1.2f%%',
       startangle=40)
plt.title('Pie chart')
plt.savefig('Continent_Pie.jpg')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Columns to visualize
cols = ['continent', 'case_status']

# Define colors for each column
colors = {
    'continent': ['goldenrod', 'blue', 'red', 'yellow', 'pink', 'purple'],
    'case_status': ['seagreen', 'salmon']
}

# Create figure
fig, axes = plt.subplots(1, 2, figsize=(18, 5))

# Loop through each column
for ax, col in zip(axes, cols):

    counts = visa[col].value_counts()

    ax.bar(
        counts.index,
        counts.values,
        color=colors[col],
        edgecolor='black'
    )

    ax.set_title(f'Distribution of {col.title()}', fontsize=14, fontweight='bold')
    ax.set_xlabel(col.title(), fontsize=12)
    ax.set_ylabel('Count', fontsize=12)
    ax.tick_params(axis='x', rotation=15)

plt.tight_layout()
plt.show()
```

```python
cols=['continent','case_status']

plt.figure(figsize=(18,3))
for i in  range(len(cols)):
    c=cols[i]
    keys=visa[c].value_counts().keys()
    values=visa[c].value_counts().values
    colors=['seagreen','salmon','red','Yellow','Pink','Purple']

    plt.subplot(1,2,i+1).bar(keys,values,
                            color=colors[:len(keys)])
    plt.title(c)
# plt.tight_layout()
plt.savefig('Continent-Case_status_bar.jpg')
plt.show()
```
